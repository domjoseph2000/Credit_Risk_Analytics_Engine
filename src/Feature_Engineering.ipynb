```python
def get_last_approved(x, full_df):
    approved = full_df.loc[x.index, 'NAME_CONTRACT_STATUS'] == 'Approved'
    return x[approved].iloc[0] if approved.any() else np.nan

def get_last_refused(x, full_df):
    refused = full_df.loc[x.index, 'NAME_CONTRACT_STATUS'] == 'Refused'
    return x[refused].iloc[0] if refused.any() else np.nan
```

```python
def get_days_to_first_dpd(x):
    
    months_bal = CCBal.loc[x.index][CCBal.loc[x.index,'SK_DPD']>=7]
    if months_bal.shape[0] > 0:
        return months_bal['MONTHS_BALANCE'].min() * -30
    else:
        return np.nan
```
