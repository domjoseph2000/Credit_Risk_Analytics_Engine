```python
PrevLoans['CREDIT_TO_APPL'] = np.where( PrevLoans['AMT_APPLICATION'] > 0,
            PrevLoans['AMT_CREDIT'] / PrevLoans['AMT_APPLICATION'],
            np.nan)

PrevLoans['GOODSPRICE_TO_AMT'] = np.where( PrevLoans['AMT_APPLICATION'] > 0,
            PrevLoans['AMT_GOODS_PRICE'] / PrevLoans['AMT_APPLICATION'],
            np.nan)
```

```python
ApprovedLoans = PrevLoans[~ (PrevLoans['NAME_CONTRACT_STATUS'] == 'Refused')]
RefusedLoans  = PrevLoans[PrevLoans['NAME_CONTRACT_STATUS'] == 'Refused']
```

```python
RefusedLoans_Silver = RefusedLoans.groupby('SK_ID_CURR').agg(
    PREVIOUS_LOAN_REFUSED_COUNT = ('NAME_CONTRACT_STATUS',lambda x: sum(x=='Refused')),
    PREVIOUS_LOAN_REFUSED_AVG_LOAN_AMOUNT = ('AMT_CREDIT','mean'),
    PREVIOUS_LOAN_REFUSED_AVG_GOODS_TO_AMT_RATIO = ('GOODSPRICE_TO_AMT','mean'),
    PREVIOUS_LOAN_REFUSED_LASTYEAR_COUNT = ('DAYS_DECISION', lambda x: sum(x > -365)),
    PREVIOUS_LOAN_LAST_REFUSED_DAY = ('DAYS_DECISION', 'max'),
    PREVIOUS_LOAN_REFUSED_MULTIPLE_ATTEMPTS_COUNT = ('FLAG_LAST_APPL_PER_CONTRACT', lambda x: sum(x == 'N'))
)
```

```python
def get_last_approved(x, full_df):
    approved = full_df.loc[x.index, 'NAME_CONTRACT_STATUS'] == 'Approved'
    return x[approved].iloc[0] if approved.any() else np.nan
```

```python
ApprovedLoans_Silver = ApprovedLoans.groupby('SK_ID_CURR').agg(

    PREVIOUS_LOAN_APPROVED_COUNT = ('NAME_CONTRACT_STATUS',lambda x: sum(x=='Approved')),
    PREVIOUS_LOAN_CANCELED_COUNT = ('NAME_CONTRACT_STATUS',lambda x: sum(x=='Canceled')),
    
    PREVIOUS_LOAN_AVG_AMOUNT = ('AMT_CREDIT','mean'),
    PREVIOUS_LOAN_MAX_AMOUNT = ('AMT_CREDIT','max'),
    PREVIOUS_LOAN_TERM_AVG = ('CNT_PAYMENT','mean'),
    PREVIOUS_LOAN_TERM_MAX = ('CNT_PAYMENT','max'),

    PREVIOUS_LOAN_AVG_DOWNPAYMENT_AMOUNT = ('AMT_DOWN_PAYMENT','mean'),
    PREVIOUS_LOAN_AVG_DOWNPAYMENT_RATIO = ('RATE_DOWN_PAYMENT','mean'),
    PREVIOUS_LOAN_AVG_CREDIT_TO_APPLIED_RATIO = ('CREDIT_TO_APPL', 'mean'),

    PREVIOUS_LOAN_NUM_OF_HIGHYIELD_LOANS = ('NAME_YIELD_GROUP', lambda x: sum(x == 'high')),

    PREVIOUS_LOAN_NUM_OF_CASHLOANS = ('NAME_CONTRACT_TYPE', lambda x: sum(x == 'Cash loans')),
    PREVIOUS_LOAN_NUM_OF_CONSUMERLOANS = ('NAME_CONTRACT_TYPE', lambda x: sum(x == 'Consumer loans')),
    PREVIOUS_LOAN_NUM_OF_REVOLVINGLOANS = ('NAME_CONTRACT_TYPE', lambda x: sum(x == 'Revolving loans')),
    
    PREVIOUS_LOAN_NUM_OF_HIGHRISK_CHANNELS = ('CHANNEL_TYPE', lambda x: sum( (x == 'AP+ (Cash loan)') | (x == 'Car dealer'))),

    PREVIOUS_LOAN_INSURANCE_REQUESTED = ('NFLAG_INSURED_ON_APPROVAL','max'),
    PREVIOUS_LOAN_NUM_OF_INSURANCE = ('NFLAG_INSURED_ON_APPROVAL','count'),

    PREVIOUS_LOAN_DAYS_BETWEEN_LOANS=('DAYS_DECISION', lambda x: x.diff().mean()),
    PREVIOUS_LOAN_DAYS_LAST_APPROVED=('DAYS_DECISION', lambda x: get_last_approved(x, ApprovedLoans)),
    PREVIOUS_LOAN_LAST_LOAN_STATUS=('NAME_CONTRACT_STATUS', 'last'),
    PREVIOUS_LOAN_MULTIPLE_ATTEMPTS_COUNT=('FLAG_LAST_APPL_PER_CONTRACT', lambda x: sum(x == 'N'))
)
```

```python
PrevLoans_Silver = PrevLoans.groupby('SK_ID_CURR').agg(
    PREVIOUS_LOAN_APPL_COUNT = ('SK_ID_PREV','count')
    ).merge(
        ApprovedLoans_Silver,
        on='SK_ID_CURR',
        how='left'
    ).merge(
        RefusedLoans_Silver,
        on='SK_ID_CURR',
        how='left'
    )
```

```python
PrevLoans_Silver['PREVIOUS_LOAN_REFUSED_COUNT'].fillna(0, inplace=True)
PrevLoans_Silver['PREVIOUS_LOAN_REFUSED_AVG_LOAN_AMOUNT'].fillna(0, inplace=True) 
PrevLoans_Silver['PREVIOUS_LOAN_REFUSED_AVG_GOODS_TO_AMT_RATIO'].fillna(0, inplace=True)
PrevLoans_Silver['PREVIOUS_LOAN_REFUSED_LASTYEAR_COUNT'].fillna(0, inplace=True)
#PrevLoans_Silver['PREVIOUS_LOAN_LAST_REFUSED_DAY'].fillna(10000 inplace=True)
PrevLoans_Silver['PREVIOUS_LOAN_REFUSED_MULTIPLE_ATTEMPTS_COUNT'].fillna(0, inplace=True)
```
