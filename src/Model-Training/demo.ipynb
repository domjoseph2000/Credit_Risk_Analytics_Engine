```python
import lightgbm as lgb
import numpy as np

# 1. Generate some random dummy data
print("Generating dummy data...")
X = np.random.rand(10000, 50)
y = np.random.randint(0, 2, 10000)
train_data = lgb.Dataset(X, label=y)

# 2. Configure the parameters to trigger the GPU
params = {
    'objective': 'binary',
    'device': 'cpu',       # This is the magic parameter
 #   'gpu_platform_id': 0,  # Tells it to use the primary OpenCL platform
 #   'gpu_device_id': 0,    # Tells it to use the primary GPU
    'verbosity': 2           # Verbose mode will print the GPU detection logs
}

# 3. Train a quick model
try:
    print("Starting GPU training...")
    bst = lgb.train(params, train_data, num_boost_round=5)
    print("\n✅ SUCCESS! LightGBM trained the model using your GPU.")
except Exception as e:
    print("\n❌ FAILED. LightGBM could not use the GPU. Error details:")
    print(e)
```
